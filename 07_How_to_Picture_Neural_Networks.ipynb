# **Correlation Summarization**

*Correlation summarization* adalah konsep tingkat tinggi (*10,000-foot view*) bahwa seluruh operasi jaringan saraf (nodus, gradien, penumpukan layer) pada hakikatnya bertujuan untuk mencari dan menciptakan korelasi antara layer input dan layer output.
* **Local Correlation Summarization**: Setiap matriks bobot mengoptimalkan nilainya untuk mengorelasikan masukan layer terkait dengan target sinyal yang diminta oleh layer berikutnya.
* **Global Correlation Summarization**: Jaringan secara keseluruhan mencari korelasi langsung (*direct*) dan tidak langsung (*indirect*) antara dataset input dan dataset output.

# **The Simplified Visualization**

Mengabstraksikan komponen jaringan saraf seperti blok LEGO:

* Matriks (diwakili oleh kotak) untuk menyatakan sekumpulan bobot ($W_0, W_1$)
